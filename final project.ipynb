```python
df=df.dropna(subset=["id kharidar"])
```

```python
df.isnull().sum()
```

```python
df["gheimat vahed"]=df["gheimat vahed"].fillna(df["gheimat vahed"].mean())
df.isnull().sum()
```

```python
df.shape
```

```python
df["revenue"]=df["tedad"]*df["gheimat vahed"]
df.head(5)
```

```python
product=df.groupby("tozihat")["tedad"].sum()
print(product)
```

```python
product=product.sort_values(ascending=False).head(10)
```

```python
print (product)
```

```python
revenue=df.groupby("tozihat")["revenue"].sum()
revenue=revenue.sort_values(ascending=False).head(10)
print (revenue)
```

```python
product=pd.DataFrame(product)
sns.set_theme(font_scale=0.8)
sns.barplot(data=product, x="tozihat", y="tedad",)
plt.title("Top 10 Products by Quantity Sold")
plt.xlabel("product type")
plt.ylabel("Quantity Sold")
plt.xticks(rotation=80)
plt.show()
```
